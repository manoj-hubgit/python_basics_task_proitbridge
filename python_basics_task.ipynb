## Task 3
### Rock, Paper, Scissors Game

```python
import random

def start_play():
    user_choise = input("Enter rock, paper or scissors: ").lower()
    valid_choise = ["rock", "paper", "scissors"]
    system_choise = random.choice(valid_choise)

    if user_choise not in valid_choise:
        print("Invalid Choise")
        return True

    print("Your choice:", user_choise)
    print("My choice:", system_choise)

    if user_choise == system_choise:
        print("It's a Tie")

    elif user_choise == "rock" and system_choise == "scissors":
        print("You won... 🎉🎉")

    elif user_choise == "paper" and system_choise == "rock":
        print("You won... 🎉🎉")

    elif user_choise == "scissors" and system_choise == "paper":
        print("You won... 🎉🎉")

    else:
        print("I Won ... 😁😁")

    play = input("Do You want to play? Enter y to play or c to quit: ")
    if play == "y":
        return True
    else:
        return False

play = True
while play:
    play = start_play()

print("Thanks for playing!")
```

## Task 4
### Word Guessing Game (Mini Hangman)

```python

import random

words = ["python", "machine", "learning", "bridge"]
answer = random.choice(words)
word_count = ["_"] * len(answer)

correct_count = 0
wrong_count = 0

print(" ".join(word_count))

while wrong_count < 6 and correct_count < len(answer):
    user_input = input("Enter a letter: ").lower()
    if user_input in answer:
        for i in range(len(answer)):
            if answer[i] == user_input and word_count[i] == "_":
                word_count[i] = user_input
                correct_count += 1
        print("Correct guess!")
    else:
        wrong_count += 1
        print("Wrong guess!")
        print("Remaining chances:", 6 - wrong_count)

    print(" ".join(word_count))

if correct_count == len(answer):
    print("You guessed the right word:", answer)

else:
    print("Better luck next time! The right answer is:", answer)
```
